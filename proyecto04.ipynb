```python
import numpy as np
import pandas as pd
from pathlib import Path

# Crear nuevamente un archivo Excel amplio y reproducible
archivo = Path('datos_ejemplo.xlsx')
rng = np.random.default_rng(42)
registros = 1000

productos = [
    'Laptop Pro 14', 'Monitor 27', 'Teclado mecanico', 'Mouse inalambrico',
    'Auriculares Bluetooth', 'Tablet Plus', 'Camara web HD', 'Disco SSD 1TB',
    'Impresora multifuncion', 'Silla ergonomica', 'Router WiFi 6', 'Smartwatch'
]
categorias = ['Tecnologia', 'Accesorios', 'Oficina', 'Movilidad']
regiones = ['Norte', 'Sur', 'Este', 'Oeste', 'Centro']
metodos_pago = ['Tarjeta', 'Efectivo', 'Transferencia', 'PayPal']

fecha = pd.Timestamp('2024-01-01') + pd.to_timedelta(rng.integers(0, 366, registros), unit='D')
producto = rng.choice(productos, registros)
categoria = rng.choice(categorias, registros, p=[0.35, 0.30, 0.20, 0.15])
region = rng.choice(regiones, registros)
unidades = rng.integers(1, 12, registros)
clientes = rng.integers(1, 30, registros)
precio_unitario = rng.uniform(15, 650, registros).round(2)
ventas = (unidades * precio_unitario).round(2)
descuentos = (ventas * rng.uniform(0.00, 0.18, registros)).round(2)
costo = (ventas * rng.uniform(0.42, 0.78, registros)).round(2)
ganancia = (ventas - descuentos - costo).round(2)
margen = (ganancia / ventas).round(4)
ticket_promedio = (ventas / clientes).round(2)

df = pd.DataFrame({
    'Fecha': fecha,
    'Producto': producto,
    'Ventas': ventas,
    'Unidades': unidades,
    'Categoria': categoria,
    'Region': region,
    'Clientes': clientes,
    'Metodo_Pago': rng.choice(metodos_pago, registros),
    'Costo': costo,
    'Ganancia': ganancia,
    'Margen': margen,
    'Descuentos': descuentos,
    'Ticket_Promedio': ticket_promedio,
}).sort_values('Fecha').reset_index(drop=True)

df.to_excel(archivo, index=False)
print(f'Archivo recreado: {archivo.resolve()}')
print(f'Registros creados: {len(df)}')
print(f'Productos distintos: {df["Producto"].nunique()}')
print(df.head())
print(f'Total de ventas: {df["Ventas"].sum():,.2f}')
print(f'Promedio de ventas: {df["Ventas"].mean():,.2f}')
```

```python
# ETAPA 1 — CARGA Y REVISIÓN INICIAL

## Objetivo
# Se valida la estructura real del archivo Excel antes de analizarlo.

import pandas as pd
import numpy as np
from pathlib import Path

archivo = Path('datos_ejemplo.xlsx')

if not archivo.exists():
    raise FileNotFoundError(
        f'No se encontró el archivo {archivo}. Colócalo en la misma carpeta del notebook antes de ejecutar el análisis.'
    )

# Carga del archivo
ventas = pd.read_excel(archivo)

print('=== REVISIÓN INICIAL ===')
print(f'Dimensiones del dataset: {ventas.shape}')
print('\nColumnas:')
print(list(ventas.columns))
print('\nTipos de datos:')
print(ventas.dtypes)
print('\nValores nulos:')
print(ventas.isnull().sum())
print('\nDuplicados:')
print(f'Filas duplicadas: {ventas.duplicated().sum()}')
print('\nValores únicos por columnas categóricas:')
for col in ventas.select_dtypes(include=['object', 'category']).columns:
    print(f'- {col}: {ventas[col].dropna().unique()[:10].tolist()}')
print('\nEstadísticas descriptivas:')
print(ventas.describe(include='all').T)
```

## ETAPA 2 — LIMPIEZA Y TRANSFORMACIÓN

En esta etapa se revisa el formato de cada variable y se crea una copia limpia para el análisis. La conversión de `Fecha` a datetime y la normalización de columnas numéricas es necesaria para poder calcular ventas, márgenes y series temporales sin errores de tipo.


```python
# ETAPA 2 — LIMPIEZA Y TRANSFORMACIÓN

import pandas as pd
import numpy as np

# Copia limpia del dataset para no alterar los datos originales
ventas_clean = ventas.copy()

# 1) Fecha: convertir a datetime
# Esto permite trabajar con series temporales y agrupar por mes o día correctamente.
ventas_clean['Fecha'] = pd.to_datetime(ventas_clean['Fecha'])

# 2) Variables numéricas: forzar tipos adecuados
# Se corrigen columnas que deben ser numéricas para evitar errores en sumas, promedios y comparaciones.
for col in ['Ventas', 'Unidades', 'Clientes', 'Costo', 'Ganancia', 'Descuentos', 'Ticket_Promedio']:
    ventas_clean[col] = pd.to_numeric(ventas_clean[col], errors='coerce')

# 3) Margen: asegurar que esté en rango lógico
# Un margen fuera de 0 a 1 puede indicar dato inconsistente.
ventas_clean['Margen'] = pd.to_numeric(ventas_clean['Margen'], errors='coerce')
ventas_clean['Margen'] = ventas_clean['Margen'].clip(lower=0, upper=1)

# 4) Revisión de nulos y duplicados
print('Valores nulos tras la limpieza:')
print(ventas_clean.isnull().sum())
print('\nDuplicados tras la limpieza:')
print(ventas_clean.duplicated().sum())

# 5) Revisar valores negativos en columnas clave
print('\nValores negativos en columnas clave:')
for col in ['Ventas', 'Unidades', 'Costo', 'Ganancia', 'Descuentos', 'Ticket_Promedio']:
    print(f'- {col}: {(ventas_clean[col] < 0).sum()}')

# 6) Ver datos finales limpios
print('\nVista general del dataset limpio:')
print(ventas_clean.head())
```

# ETAPA 3 — DOCUMENTACIÓN DEL EDA

## Análisis Exploratorio de Datos — Ventas 2024

### 1. Objetivo del análisis
Este notebook tiene como objetivo revisar la calidad de los datos, limpiar la estructura, analizar el comportamiento de las ventas y evaluar la rentabilidad del negocio para 2024.

Se trabaja con un archivo Excel de ventas diarias con columnas como Fecha, Ventas, Unidades, Categoria, Region, Clientes, Metodo_Pago, Costo, Ganancia, Margen, Descuentos y Ticket_Promedio.

### 3. Calidad y estructura de los datos
Se verificó dimensiones, tipos, nulos, duplicados y variables categóricas. La estructura inicial del archivo fue consistente y no se detectaron valores faltantes ni duplicados.

### 4. Limpieza y transformación
Se convirtió la columna Fecha a tipo datetime y se normalizaron las columnas numéricas para garantizar cálculos fiables. El margen se limitó a un rango lógico entre 0 y 1.

### 5. Análisis descriptivo
Se utilizarán métricas como total de ventas, promedio diario, unidades, costo, ganancia, margen y ticket promedio para resumir el comportamiento del negocio.

### 6. Análisis de ventas

```python
# ETAPA 4 — ANÁLISIS GENERAL

# Resumen financiero y operativo
ventas_summary = {
    'Ventas_totales': ventas_clean['Ventas'].sum(),
    'Ventas_promedio_diarias': ventas_clean['Ventas'].mean(),
    'Ventas_mediana': ventas_clean['Ventas'].median(),
    'Registros': len(ventas_clean),
    'Unidades_vendidas': ventas_clean['Unidades'].sum(),
    'Costo_total': ventas_clean['Costo'].sum(),
    'Ganancia_total': ventas_clean['Ganancia'].sum(),
    'Margen_promedio': ventas_clean['Margen'].mean(),
    'Margen_ponderado': ventas_clean['Ganancia'].sum() / ventas_clean['Ventas'].sum(),
    'Descuentos_totales': ventas_clean['Descuentos'].sum(),
    'Ticket_promedio': ventas_clean['Ticket_Promedio'].mean(),
    'Clientes_totales': ventas_clean['Clientes'].sum(),
}

print('=== RESUMEN GENERAL ===')
for k, v in ventas_summary.items():
    print(f'{k}: {v}')

# Evolución mensual
ventas_clean['Mes'] = ventas_clean['Fecha'].dt.to_period('M').astype(str)
resumen_mensual = ventas_clean.groupby('Mes').agg(
    Ventas=('Ventas', 'sum'),
    Ganancia=('Ganancia', 'sum'),
    Costo=('Costo', 'sum'),
    Unidades=('Unidades', 'sum'),
    Ticket_Promedio=('Ticket_Promedio', 'mean'),
    Descuentos=('Descuentos', 'sum')
).reset_index()
resumen_mensual['Margen'] = resumen_mensual['Ganancia'] / resumen_mensual['Ventas']

print('\n=== RESUMEN MENSUAL ===')
print(resumen_mensual)

# Rentabilidad por categoría
resumen_categoria = ventas_clean.groupby('Categoria').agg(
    Ventas=('Ventas', 'sum'),
    Ganancia=('Ganancia', 'sum'),
    Unidades=('Unidades', 'sum'),
    Clientes=('Clientes', 'sum'),
    Descuentos=('Descuentos', 'sum')
).reset_index()
resumen_categoria['Margen'] = resumen_categoria['Ganancia'] / resumen_categoria['Ventas']
print('\n=== RESUMEN POR CATEGORÍA ===')
print(resumen_categoria)

# Rentabilidad por región
resumen_region = ventas_clean.groupby('Region').agg(
    Ventas=('Ventas', 'sum'),
    Ganancia=('Ganancia', 'sum'),
    Unidades=('Unidades', 'sum'),
    Clientes=('Clientes', 'sum'),
    Descuentos=('Descuentos', 'sum')
).reset_index()
resumen_region['Margen'] = resumen_region['Ganancia'] / resumen_region['Ventas']
print('\n=== RESUMEN POR REGIÓN ===')
print(resumen_region)
```

```python
# ETAPA 5 — VISUALIZACIONES INTERACTIVAS

import plotly.express as px

# Gráfico 1: evolución mensual de ventas y ganancia
fig_ventas = px.line(
    resumen_mensual,
    x='Mes',
    y=['Ventas', 'Ganancia'],
    title='Evolución mensual de ventas y ganancia',
    labels={'value': 'Monto', 'variable': 'Indicador', 'Mes': 'Mes'}
)
fig_ventas.show()

# Gráfico 2: ventas y ganancia por categoría
fig_cat = px.bar(
    resumen_categoria,
    x='Categoria',
    y=['Ventas', 'Ganancia'],
    barmode='group',
    title='Ventas y ganancia por categoría',
    labels={'value': 'Monto', 'variable': 'Indicador'}
)
fig_cat.show()

# Gráfico 3: margen por categoría
fig_margin_cat = px.bar(
    resumen_categoria,
    x='Categoria',
    y='Margen',
    title='Rentabilidad por categoría (margen)',
    labels={'Margen': 'Margen', 'Categoria': 'Categoría'}
)
fig_margin_cat.show()

# Gráfico 4: ventas y margen por región
fig_region = px.bar(
    resumen_region,
    x='Region',
    y=['Ventas', 'Ganancia'],
    barmode='group',
    title='Ventas y ganancia por región',
    labels={'value': 'Monto', 'variable': 'Indicador'}
)
fig_region.show()

# Gráfico 5: relación entre ventas, costo y ganancia
fig_rel = px.scatter(
    ventas_clean,
    x='Ventas',
    y='Ganancia',
    size='Costo',
    color='Categoria',
    title='Relación entre ventas, costo y ganancia',
    labels={'Ventas': 'Ventas', 'Ganancia': 'Ganancia', 'Costo': 'Costo'}
)
fig_rel.show()

# Gráfico 6: impacto de descuentos sobre el margen
fig_desc = px.scatter(
    ventas_clean,
    x='Descuentos',
    y='Margen',
    color='Categoria',
    size='Ventas',
    title='Impacto de los descuentos sobre el margen',
    labels={'Descuentos': 'Descuentos', 'Margen': 'Margen', 'Ventas': 'Ventas'}
)
fig_desc.show()

# Gráfico 7: ticket promedio por categoría
fig_ticket = px.bar(
    ventas_clean.groupby('Categoria')['Ticket_Promedio'].mean().reset_index(),
    x='Categoria',
    y='Ticket_Promedio',
    title='Ticket promedio por categoría',
    labels={'Ticket_Promedio': 'Ticket promedio', 'Categoria': 'Categoría'}
)
fig_ticket.show()
```

```python
# ETAPA 6 — TABLAS DE ANÁLISIS

# Resumen mensual
print('=== RESUMEN MENSUAL ===')
print(resumen_mensual.round(2).to_string(index=False))

# Ranking por categoría
ranking_categoria = resumen_categoria.sort_values('Ventas', ascending=False).reset_index(drop=True)
print('\n=== RANKING POR CATEGORÍA ===')
print(ranking_categoria.round(2).to_string(index=False))

# Ranking por región
ranking_region = resumen_region.sort_values('Ventas', ascending=False).reset_index(drop=True)
print('\n=== RANKING POR REGIÓN ===')
print(ranking_region.round(2).to_string(index=False))

# Tabla de rentabilidad por categoría con margen y participación
categoria_rentabilidad = resumen_categoria[['Categoria', 'Ventas', 'Ganancia', 'Margen']].copy()
categoria_rentabilidad['Participacion_Ventas'] = categoria_rentabilidad['Ventas'] / categoria_rentabilidad['Ventas'].sum()
print('\n=== RENTABILIDAD POR CATEGORÍA ===')
print(categoria_rentabilidad.round(4).to_string(index=False))
```
